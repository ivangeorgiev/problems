## Discounted Cash Flow

### 1. Estimate Free Cash Flows for Each Year

* BYFCF - Base Year Free Cash Flow
* $FCF_n$ - Free cash flow for year $n$
* $n$ - the number of years into the future
* $GR$ - the company's estimated annual growth rate

$$
FCF_n = BYFCF*(1 + GR)^n
$$

### 2. Estimate the Discount Factor for Each Year

$$
DF_n = (1+DR)^n
$$

* $DF_n$ - The discount factor during year n
* $DR$ - Discount rate
* $n$ - Year being discounted

### 3. Calculate the Discount Value of FCF for N=10 Years

$$
DFCF_n = \frac{FCF_n}{DF_n} = BYFCF \left( \frac{1+GR}{1+DR} \right)
$$

* $DFCF_n$ - Discounted free cash flow for year $n$
* $FCF_n$ - Free cash flow for year $n$
* $GR$ - Estimated annual growth rate
* $DR$ - Discount rate

### Calculate the Discounted Perpetuity Free Cash Flow (beyond ten years)

$$
DPCF = BYFCF \frac{(1 + GR)^{N+1}(1+LGR)}{DR - LGR} \frac{1}{(1+DR)^{N+1}}
$$
* $DPCF$ - Discounted perpetuity cash flow
* $BYFCF$ - Base year free cash flow
* $GR$ - Growth rate of FCF
* $DR$ - Discount rate
* $LGR$ - Long-term growth rate

$$
S_n = 1 + q + ... + q^n = \frac{1-q^n}{1-q}
$$

$$
S = lim_{n \to \inf} S_n = \frac{1}{1-q}
$$

$$
PCF_n = FCF_N*(1+LGR)^n
$$

$$
DPCF_n = \frac{DFCF_N(1+LGR)^n}{(1+DR)^n} = FCF_N \left( \frac{1+LGR}{1+DR} \right)^n
$$

$$
\begin{align}
DPCF & = \sum_{n=1}^{\inf} DPCF_n   \\
  & = DFCF_N \sum_{n=1}^{\inf} \left( \frac{1+LGR}{1+DR} \right)^n  \\
  & = DFCF_N \frac{1}{1 - \frac{1+LGR}{1+DR}}  \\
  & = BYCF \frac{1+DR}{DR-LGR} \left( \frac{1+GR}{1+DR}  \right)^{N+1}
\end{align}
$$

```python
import numpy as np
import pandas as pd
```

```python
GR = 0.06
BYCF = 1000
DR = 0.1
LGR = 0.03
N = 10

DPCF = BYCF*(1+DR)/(DR-LGR)*((1+GR)/(1+DR))**(N+1)
```

## The Combined DCF Formula

The total value ($V_0$) is expressed as:
$$
V_0 = \sum_{t=1}^{n} \frac{FCF_t}{(1+r)^t} + \frac{TV_n}{(1+r)^n}
$$

Where the **Terminal Value** ($TV_n) at the end of the forecast period ($n$) is calculated using the perpetual growth method (Gordon Growth Model):

$$
TV_n = \frac{FCF_n \times (1+g)}{r-g} = \frac{FCF_{n+1}}{r-g}
$$

**Key Variables:**
* $FCF_t$: Free cash flow in explicit period $t$.
* $FCF_n$: Free cash flow in the final forecasted year.
* $FCF_{n+1}$: Projected free cash flow for the first year beyond the forecast period ($FCF_n\times (1+g)$).
* $r$: Discount rate, typically the Weighted Average Cost of Capital (WACC).
* $g$: Perpetual growth rate of the cash flows beyound year $n$.
* $n$: Number of explicit forecast periods (years).

```python
g = 2.5/100
r = 10.0/100
FCF_0 = 30.0
GR = 15.0/100
n=10
```

```python
# Free Cash Flows
FCF_t = FCF_0*(1+GR)**np.arange(1,n+1)
FCF_t
```

```python
# Discounted Free Cash Flows
DFCF_t = FCF_t/(1+r)**np.arange(1,n+1)
DFCF_t
```

```python
# Terminal Value
TV = FCF_t[-1]*(1+g)/(r-g)
TV
```

```python
# Discounted Terminal Value
DTV = TV/(1+r)**n
DTV
```

```python
# Value
V_0 = DFCF_t.sum() + DTV
V_0
```

```python
import numpy as np

def period_indexes(n) -> np.ndarray:
    """Returns array of period indexes: 1, 2, 3,..."""
    return np.arange(1, n+1)

def discounted(values, discount_rate) -> np.ndarray:
    """
    Calculate discounted present value using: Discounted_t = Value_t / (1 + discount_rate)^t
    """
    # https://medium.com/@gianlucabaglini/automate-your-discounted-cash-flow-model-in-python-cdf98eb0924d
    period_indexes = np.arange(1, len(values)+1)
    discounted_values = values/(1 + discount_rate)**period_indexes
    return discounted_values
    
def grow_by_rate(initial, rate, num_iterations):
    """Grow a metric by given rate for given number of iterations"""
    return initial*(1+rate)**np.arange(1, num_iterations+1)

def grow_by(initial, growth, num_iterations):
    """Grow a metric by adding a growth at each iteration"""
    return initial + growth*np.arange(1, num_iterations+1)

def total_value_growth_in_perpetuity(free_cash_flows, perpetual_growth_rate, discount_rate, *, output_as="number"):
    n = len(free_cash_flows)
    # Convert to np.array
    fcf_n = np.array(free_cash_flows)
    # discounted free cash flows
    dfcf_n = discounted(fcf_n, discount_rate)
    # discounted free cash flow for all years
    dfcf = dfcf_n.sum()
    # terminal value
    tv = fcf_n[-1]*(1+perpetual_growth_rate)/(discount_rate - perpetual_growth_rate)
    # discounted terminal value
    dtv = tv / (1+discount_rate)**n
    total_value = dfcf + dtv
    match output_as:
        case "number":
            return total_value
        case "dict":
            return {
                "Long-term growth rate": perpetual_growth_rate,
                "Discount rate": discount_rate,
                "Number of years": n,
                "Free Cash Flows": free_cash_flows,
                "Discounted Free Cash Flows": dfcf_n,
                "Total Discounted Free Cash Flow": dfcf,
                "Terminal Value at last year": tv,
                "Discounted Terminal Value": dtv,
                "Total Value": total_value
            }
        case _:
            raise ValueError("Unknown value for output_as. Allowed values: number, dict")
```

```python
fcf_n = grow_by_rate(30.0, 0.06, 10)
fcf_n
```

```python
tv = total_value_growth_in_perpetuity(fcf_n, 0.025, 0.10, output_as="dict")
tv
```

### Example

Based on https://www.wallstreetprep.com/knowledge/terminal-value/

In this example we assume that free cash flow grows linear with EBITDA. EBITDA grows linearly by fixed amount, e.g. $2mm.

```python
EBITDA_0 = 50  # millions, mm
EBITDA_growth = 2 # millions, mm
FCF_0 = 30   # millions, mm
n = 5 # years
ebitda_conversion_ratio = FCF_0 / EBITDA_0

r = 10.0/100.0 # discount rate
g = 2.5/100.0 # long-term growth rate

```

```python
years = np.arange(1, n+1)
years
```

```python
# EBITDA for each year grows from the previous year by EBITDA_growth
EBITDA_t = EBITDA_0 + years*EBITDA_growth
EBITDA_t
```

```python
# Free Cash Flow can be obtained from EBITDA by multiplying to the conversion ratio
FCF_t = EBITDA_t * ebitda_conversion_ratio
FCF_t
```

```python
# Can use `grow_by()` function for the future EBITDA values
grow_by(EBITDA_0, EBITDA_growth, n)*ebitda_conversion_ratio
```

```python
# Alternatively we could calculate FCF growth from EBITDA growth
FCF_growth = EBITDA_growth*FCF_0/EBITDA_0
FCF_0 + FCF_growth*years
```

```python
# Total Value in millions, mm
tv = total_value_growth_in_perpetuity(FCF_t, g, r, output_as="dict")
tv
```

```python
np.round(tv['Total Value'])
```

#### Free Cash Flow growing by rate

```python
FCF_t = grow_by_rate(FCF_0, EBITDA_growth/EBITDA_0, n)
FCF_t
```

```python
tv = total_value_growth_in_perpetuity(FCF_t, g, r, output_as="dict")
tv
```

```python
np.round(tv['Total Value'])
```
