## Present Value of a Purchase

I am buying a car today for $30000. Ho much it actually costs me.

```python
P = 30000
n = 10 # years
r = 0.1 # market rate of return
i = 0.03 # inflation
```

```python
import numpy as np
import numpy_financial as npf
```

```python
FV = npf.fv(r,n,0,P)
FV
```

```python
PV = npf.pv(i,n,0,FV)
PV
```

```python
PV/P
```

Actual value of the purchase is 58000. Almost double (1.93 times) the initial investment.

### With Fixed Periodic Maintenance Cost

Now let's assume that the car needs $1000 additional cost per year (maintenance, insurance etc.)

For the simple approach we assume that $1000 cost is fixed amount.

```python
P = 30000
p = 1000
n = 10 # years
r = 0.1 # market rate of return
i = 0.03 # inflation
```

```python
# Funture value of P:
FVP = P*(1+r)**n
FVP
```

```python
# Future value of regular payments p:
FVp = p*((1+r)**n-1)/r
FVp
```

```python
# Present value
PV = (FVP + FVp)/(1+i)**n
PV
```

Taking in account adds 12000 more (40%) to the total cost of the purchase.

### With Growing Periodic Maintenance Cost

```python
P = 30000
p = 1000
n = 10 # years
r = 0.1 # market rate of return
i = 0.03 # inflation
```

```python
# Funture value of P:
FVP = P*(1+r)**n
FVP
```

```python
# Actual inflated periodic payments
period_index = np.arange(10)
actual_p = p*np.pow(1+i, period_index)
actual_p
```

```python
# Future value of actual periodic payments
reversed_period_index = period_index[::-1]
FVp = sum(actual_p*(1+r)**reversed_period_index)
FVp
```

```python
# Present value
PV_inflated = (FVP + FVp)/(1+i)**n
PV_inflated
```

```python
PV_inflated - PV
```

```python
(PV_inflated - PV)/P*100
```

```python
round((PV_inflated - P)/P*100, 2)
```

Taking in account inflation adds another $1400 (about 4.8%).

The actual value of our car purchase is $71185 (137.28% more than current price).

### Generalization

$$
PV_{price} = P \times \left(\frac{1+r}{1+i}\right)^n
$$

$$
{actual\ payments} = p,\ p(1+i),\ ...,\ p(1+i)^{(n-1)}
$$

$$
FV_{actual\ payments} = p(1+r)^{(n-1)} + p(1+r)^{(n-2)}(1+i) + ... + p(1+i)^{(n-1)}
$$

$$
PV_{actual\ payments} = \frac{FV_{actual\ payments}}{(1+i)^n}
$$


```python
n = 10
r = 0.1
i = 0.03
periods = np.arange(n)
reverse_periods = periods[::-1]
```

```python
price_factor = (1+r)**n/(1+i)**n
price_factor = round(price_factor, 2)
price_factor
```

```python
payment_factor = ((1+i)**periods*(1+r)**reverse_periods).sum()/(1+i)**n
payment_factor = round(payment_factor, 2)
payment_factor
```

```python
30000*price_factor + 1000*payment_factor
```
