# Reverse a String

**Problem Statement** “Given a string `s`, write a function that returns the string reversed. For example, if `s = "hello"`, the output should be `"olleh"`.

## Test Cases

| Input  | Output |              |
|--------|---------------------|-------------------|
| ""   | ""   | Empty string |
| "h"   | "h"   | String with length 1 |
| "hello"   | "olleh"   | String with odd length >1 |
| "door" | "rood"   | String with even length >1 |
| "hello john" | "nhoj olleh"  | String with space |

```python
def test_reverse_string_1(fn):
    test_parameters = [
        ("", ""),
        ("h", "h"),
        ("hello", "olleh"),
        ("hello john", "nhoj olleh"),
        ("door", "rood"),
    ]
    errors = []
    for input, expected in test_parameters:
        try:
            actual = fn(input)
            assert isinstance(actual, str), f"Expected result to be `str`, found `{type(actual).__name__}`"
            assert expected == actual, f"Expected '{expected}', actual '{actual}'"
        except Exception as e:
            errors.append((input, expected, actual, str(e)))
    return errors
```

## Approach 1. Using Python Slicing

```python
def reverse_string_1(s):
    return s[::-1]
```

```python
test_reverse_string(reverse_string_1)
```

## Approach 2. Using a loop

```python
def reverse_string_2(s):
    reversed_str = ""
    for char_index in range(len(s)-1, -1, -1):
        reversed_str += s[char_index]
    return reversed_str
```

```python
test_reverse_string(reverse_string_2)
```

```python
def reverse_string_3(s):
    """
    Improved version of reverse string 2 avoding string concatenation.
    """
    reversed_chars = [s[i] for i in range(len(s)-1, -1, -1)]
    return "".join(reversed_chars)
```

```python
test_reverse_string(reverse_string_3)
```

```python
def reverse_string_4(s):
    """
    Two-pointer approach.
    """
    if len(s) < 2:
        return s
    left, right = 0, len(s) - 1
    reversed_chars = list(s)
    while left < right:
        reversed_chars[left], reversed_chars[right] = reversed_chars[right], reversed_chars[left]
        left += 1
        right -= 1
    return "".join(reversed_chars)
```

```python
test_reverse_string(reverse_string_4)
```

```python
def reverse_string_5(s):
    """
    using reversed() python function
    """
    return "".join(reversed(s))
```

```python
test_reverse_string(reverse_string_5)
```
