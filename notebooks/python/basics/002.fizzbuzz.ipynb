# FizzBuzz

**Problem statement:** “Write a program that prints the numbers from 1 to 100. However, for multiples of 3, print _Fizz_ instead of the number, and for multiples of 5, print _Buzz_. For numbers that are multiples of both 3 and 5, print _FizzBuzz_.”

```python
FIZZBUZZ_TEST_PARAMS = [
    (0, []),
    (1, [1]),
    (2, [1,2]),
    (3, [1,2, "Fizz"]),
    (
        5, [
             1,
             2,
             'Fizz',
             4,
             'Buzz',
        ]
    ),
    (
        15, [
             1,
             2,
             'Fizz',
             4,
             'Buzz',
             'Fizz',
             7,
             8,
             'Fizz',
             'Buzz',
             11,
             'Fizz',
             13,
             14,
             'FizzBuzz'
        ]
    )
]

def test_fizzbuzz(fn):
    errors = []
    for input, expected in FIZZBUZZ_TEST_PARAMS:
        actual = fn(input)
        try:
            assert actual == expected, f"Expected result `{expected}`, actual result `{actual}`"
        except Exception as e:
            errors.append( (input, expected, actual, f"{type(e)}: {str(e)}"))
    return errors
```

```python
def fizzbuzz_1(n):
    result = []
    for i in range(1, n+1):
        output = ""
        if i % 3 == 0:
            output += "Fizz"
        if i % 5 == 0:
            output += "Buzz"
        result.append(output or i)
    return result
```

```python
test_fizzbuzz(fizzbuzz_1)
```

```python
def fizzbuzz_2(n):
    """
    without if conditional operator - with ternary operator
    """
    divisibles_map = {
        15: "FizzBuzz",
        3: "Fizz",
        5: "Buzz"
    }
    if_map = {
        False: None
    }
    result = []
    for i in range(1, n+1):
        this_output = None
        for divisor, s in divisibles_map.items():
            this_output = this_output or (s if i % divisor == 0 else None)
        result.append(this_output or i)
    return result
        
```

```python
test_fizzbuzz(fizzbuzz_2)
```

```python
def fizzbuzz_3(n):
    """
    without if conditional operator,
    without ternary operator
    """
    divisibles_map = {
        15: "FizzBuzz",
        3: "Fizz",
        5: "Buzz"
    }
    if_map = {
        False: None
    }
    result = []
    for i in range(1, n+1):
        this_output = None
        for divisor, s in divisibles_map.items():
            this_output = this_output or if_map.get(i % divisor == 0, s)
        result.append(this_output or i)
    return result
        
```

```python
test_fizzbuzz(fizzbuzz_3)
```

```python
def fizzbuzz_4(n):
    """
    without conditionals
    """
    result = []
    for i in range(1, n+1):
        result.append(("Fizz"*(i%3==0) + "Buzz"*(i%5==0)) or i)
    return result
        
```

```python
test_fizzbuzz(fizzbuzz_4)
```

```python
def fizzbuzz_5(n):
    divisibles_map = {
        15: "FizzBuzz",
        3: "Fizz",
        5: "Buzz",
    }
    result = []
    for i in range(1, n+1):
        this_output = []
        for divisor, s in divisibles_map.items():
            this_output.append((i%divisor==0, s))
        this_output.append( (True, i))
        result.append(next(filter(lambda x: x[0], this_output))[1])
    return result
        
```

```python
test_fizzbuzz(fizzbuzz_5)
```
