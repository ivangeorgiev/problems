# Anagrams

**Problem Statement:** Check if one given string could be made from another by using the same letters in any orders. Punctuation, spaces, and uppercase or lowercase differences are ignored. Such strings are called anagrams.

**Examples:**

* "Listen" and "Silent" are anagrams.
* "School master" and "The classroom" are anagrams.
* "I am Lord Voldemort" and "Tom Marvolo Riddle" are anagrams.
* "Hello" and "World" are not anagrams.



```python
ANAGRAM_TESTS = [
    ("", "", True),
    ("", "world", False),
    ("Hello", "", False),
    ("bob", "rob", False),
    ("bob", "bob", True),
    ("bob", "bob12", True),
    ("I am Lord Voldemort", "Tom Marvolo Riddle", True),
]
```

```python
def test_check_anagram_func(fn):
    errors = []
    for left, right, expected in ANAGRAM_TESTS:
        try:
            actual = fn(left, right)
            assert actual == expected, f"Expected `{expected}`, but found `{actual}`" 
        except Exception as e:
            errors.append((left, right, str(e)))
        return errors
```

```python
def check_anagrams(left: str, right: str) -> bool:
    def clean(s: str):
        return [ch for ch in s.lower() if ch.isalpha()]
    return sorted(clean(left)) == sorted(clean(right))
```

```python
test_check_anagram_func(check_anagrams)
```

```python
from collections import defaultdict

def check_anagrams_2(left: str, right: str) -> bool:
    def count(a, counts, increment):
        for ch in a:
            if not ch.isalpha():
                continue
            counts[ch.lower()] += increment
        
    counts = defaultdict(lambda: 0)
    count(left, counts, 1)
    count(right, counts, -1)
    return all(not c for c in counts.values())
```

```python
test_check_anagram_func(check_anagrams_2)
```
