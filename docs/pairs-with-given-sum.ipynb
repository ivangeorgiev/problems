# Pairs with Given Sum

Find all pairs of elements in an array that sum up to a given target. Each pair should only be listed once.

```python
def find_pairs_with_sum(arr: list, target):
    seen = set()
    pairs = set()

    for x in arr:
        y = target - x
        if y in seen:
            pairs.add(tuple(sorted((x,y))))
        seen.add(x)
    return pairs
```

```python
find_pairs_with_sum([1, -1, 3, 5, 6, 7, 9, -1], 6)
```
