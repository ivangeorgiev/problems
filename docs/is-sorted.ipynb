# Is Sorted

The "Is Sorted" problem involves checking whether a given list (or array)
is sorted in either ascending or descending order. This can be useful for
validating input data or optimizing algorithms that assume sorted input.

```python
def is_sorted(arr) -> bool:
    is_ascending = True
    is_descending = True

    for i in range(1, len(arr)):
        if arr[i] > arr[i-1]:
            is_descending = False
        if arr[i] < arr[i-1]:
            is_ascending = False
    return is_ascending or is_descending
```

```python
is_sorted([1,2,3,4,5])
```

```python
is_sorted([6,5,4,3,2,1])
```

```python
is_sorted([1,3,5,2,6,4])
```

```python
is_sorted([])
```

```python
is_sorted([3])
```
