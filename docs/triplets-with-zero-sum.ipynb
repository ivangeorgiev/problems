# Find Triplets with Zero Sum

Find all unique triplets in an array that sum to zero.
Given an array of integers, the task is to find triplets (three numbers) whose
sum is zero. These triplets should be distinct, meaning the same
combination of numbers should not appear more than once.

```python
def find_triplets_with_zero_sum_1(arr) -> list[tuple]:
    triplets = set()
    for i in range(len(arr)-2):
        x = arr[i]
        for j in range(i+1, len(arr)-1):
            y = arr[j]
            for k in range(j+1, len(arr)):
                z = arr[k]
                triplet = tuple(sorted((x, y, z)))
                if sum(triplet) == 0:
                    triplets.add(triplet)
    return list(triplets)
                    
```

```python
find_triplets_with_zero_sum_1([-1, 0, 1, 2, -1, -4])
```

```python
def three_sum_zero(arr) -> list[tuple]:
    arr = sorted(arr)
    triplets = []

    for i in range(len(arr) - 2):
        # Skip adjacent duplicate values at i
        if i > 0 and arr[i] == arr[i - 1]:
            continue

        left, right = i + 1, len(arr) - 1

        while left < right:
            s = arr[i] + arr[left] + arr[right]

            if s == 0:
                triplets.append((arr[i], arr[left], arr[right]))
                left += 1
                right -= 1

                # Skip adjacent duplicates at left
                while left < right and arr[left] == arr[left - 1]:
                    left += 1

                # Skip adjacent duplicates at right
                while left < right and arr[right] == arr[right + 1]:
                    right -= 1

            elif s < 0:
                left += 1
            else:
                right -= 1

    return triplets
```
